```python
## Hands-On Example

# Let's implement a simple optimization step-by-step:

# Step 1: Define the problem
print("Step 1: Problem Definition")
print("=" * 40)

# Parameter bounds
bounds = torch.tensor([
    [0.1, 1.0],      # concentration
    [10.0, 100.0],   # print_speed
    [0.05, 0.5],     # gap_size
    [5.0, 25.0]      # volume
]).T

param_names = ['concentration', 'print_speed', 'gap_size', 'volume']
param_units = ['', 'mm/s', 'mm', 'μL']

print("Parameter space:")
for i, (name, unit) in enumerate(zip(param_names, param_units)):
    lower, upper = bounds[0, i], bounds[1, i]
    unit_str = f" ({unit})" if unit else ""
    print(f"  {name}: [{lower:.2f}, {upper:.2f}]{unit_str}")

print(f"\nTotal search space size: {np.prod(bounds[1] - bounds[0]):.2e}")
print("This is a 4-dimensional continuous optimization problem!")
```
